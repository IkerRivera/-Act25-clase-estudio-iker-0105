```python
# =========================================
# SISTEMA IA DE MUSICOS
# LISTA DE DICCIONARIOS
# Nivel: Principiante
# =========================================

# Lista vacia
musicos = []


# =========================================
# FUNCION AGREGAR MUSICO
# =========================================
def agregar_musico():

    print("\n========== REGISTRO DE MUSICO ==========")

    # Diccionario vacio
    musico = {}

    musico["Nombre"] = input("Nombre del musico: ")
    musico["Instrumento"] = input("Instrumento principal: ")
    musico["Anios"] = input("Años de estudio: ")
    musico["Banda"] = input("Orquesta o banda: ")
    musico["Partituras"] = input("Partituras compuestas: ")
    musico["Conciertos"] = input("Conciertos al año: ")
    musico["Conservatorio"] = input("Conservatorio: ")

    musicos.append(musico)

    print("\nIA: Musico registrado correctamente")


# =========================================
# FUNCION VER MUSICOS
# =========================================
def ver_musicos():

    print("\n==================== LISTA DE MUSICOS ====================")

    if musicos == []:

        print("\nNo hay musicos registrados")

    else:

        print("--------------------------------------------------------------------------------------------------------------")
        print("NOMBRE        | INSTRUMENTO    | AÑOS   | BANDA            | PARTITURAS | CONCIERTOS | CONSERVATORIO")
        print("--------------------------------------------------------------------------------------------------------------")

        for musico in musicos:

            print(
                f"{musico['Nombre']:<13} | "
                f"{musico['Instrumento']:<14} | "
                f"{musico['Anios']:<6} | "
                f"{musico['Banda']:<16} | "
                f"{musico['Partituras']:<11} | "
                f"{musico['Conciertos']:<11} | "
                f"{musico['Conservatorio']}"
            )

        print("--------------------------------------------------------------------------------------------------------------")


# =========================================
# FUNCION ANALISIS IA
# =========================================
def analisis_musical():

    print("\n========== ANALISIS MUSICAL IA ==========")

    conciertos = int(input("Cantidad de conciertos al año: "))
    anios = int(input("Años de estudio: "))

    if conciertos > 50 and anios > 10:

        print("\nIA: MUSICO PROFESIONAL")

    elif conciertos > 20 and anios > 5:

        print("\nIA: MUSICO AVANZADO")

    elif conciertos > 5 and anios > 2:

        print("\nIA: MUSICO EN DESARROLLO")

    else:

        print("\nIA: MUSICO PRINCIPIANTE")


# =========================================
# FUNCION BUSCAR MUSICO
# =========================================
def buscar_musico():

    print("\n========== BUSCAR MUSICO ==========")

    nombre = input("Nombre del musico: ")

    encontrado = False

    for musico in musicos:

        if musico["Nombre"].lower() == nombre.lower():

            print("\nMusico encontrado\n")

            for key, value in musico.items():

                print(key, ":", value)

            encontrado = True

    if encontrado == False:

        print("\nMusico no encontrado")


# =========================================
# MENU PRINCIPAL
# =========================================
while True:

    print("\n========================================")
    print("        SISTEMA IA DE MUSICOS")
    print("========================================")

    print("1. Registrar musico")
    print("2. Ver musicos")
    print("3. Analisis musical IA")
    print("4. Buscar musico")
    print("5. Salir")

    opcion = input("\nSelecciona una opcion: ")

    # =====================================
    # OPCION 1
    # =====================================

    if opcion == "1":

        agregar_musico()

    # =====================================
    # OPCION 2
    # =====================================

    elif opcion == "2":

        ver_musicos()

    # =====================================
    # OPCION 3
    # =====================================

    elif opcion == "3":

        analisis_musical()

    # =====================================
    # OPCION 4
    # =====================================

    elif opcion == "4":

        buscar_musico()

    # =====================================
    # OPCION 5
    # =====================================

    elif opcion == "5":

        print("\nSistema finalizado")
        break

    # =====================================
    # OPCION INVALIDA
    # =====================================

    else:

        print("\nOpcion no valida")
```
