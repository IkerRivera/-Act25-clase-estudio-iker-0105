## rol : Investigador: Línea de investigación, publicaciones científicas, universidad, becas obtenidas, nivel de SNI y laboratorio.

```python
# =========================================
#   SISTEMA DE INVESTIGADORES
#   Diccionarios utilizando funciones
# =========================================

# Diccionario vacío (lista de diccionarios)
diccionario_investigadores = []

# -----------------------------------------
# Función para agregar investigadores
# -----------------------------------------
def agregar_investigador():
    print("\n--- AGREGAR INVESTIGADOR ---")

    id_investigador = len(diccionario_investigadores) + 1
    nombre = input("Nombre del investigador: ")
    linea = input("Línea de investigación: ")
    publicaciones = input("Publicaciones científicas: ")
    universidad = input("Universidad: ")
    becas = input("Becas obtenidas: ")
    sni = input("Nivel de SNI: ")
    laboratorio = input("Laboratorio: ")

    investigador = {
        "ID": id_investigador,
        "Nombre": nombre,
        "Linea": linea,
        "Publicaciones": publicaciones,
        "Universidad": universidad,
        "Becas": becas,
        "SNI": sni,
        "Laboratorio": laboratorio
    }

    diccionario_investigadores.append(investigador)

    print("\n✅ Investigador agregado correctamente.")


# -----------------------------------------
# Función para listar investigadores
# -----------------------------------------
def listar_investigadores():

    print("\n================ TABLA DE INVESTIGADORES ================\n")

    if len(diccionario_investigadores) == 0:
        print("No hay investigadores registrados.")

    else:

        print("{:<4} {:<20} {:<20} {:<18} {:<18} {:<15} {:<10} {:<15}".format(
            "ID",
            "NOMBRE",
            "LINEA",
            "PUBLICACIONES",
            "UNIVERSIDAD",
            "BECAS",
            "SNI",
            "LABORATORIO"
        ))

        print("-" * 130)

        for investigador in diccionario_investigadores:

            for key, value in investigador.items():
                pass

            print("{:<4} {:<20} {:<20} {:<18} {:<18} {:<15} {:<10} {:<15}".format(
                investigador["ID"],
                investigador["Nombre"],
                investigador["Linea"],
                investigador["Publicaciones"],
                investigador["Universidad"],
                investigador["Becas"],
                investigador["SNI"],
                investigador["Laboratorio"]
            ))


# -----------------------------------------
# Función para buscar investigador
# -----------------------------------------
def buscar_investigador():

    listar_investigadores()

    print("\n--- BUSCAR INVESTIGADOR ---")

    nombre_buscar = input("Ingresa el nombre del investigador: ")

    encontrado = False

    for investigador in diccionario_investigadores:

        if investigador["Nombre"].lower() == nombre_buscar.lower():

            print("\n=========== INVESTIGADOR ENCONTRADO ===========\n")

            for key, value in investigador.items():
                print(f"{key}: {value}")

            encontrado = True

    if encontrado == False:
        print("\n❌ Investigador no encontrado.")


# -----------------------------------------
# Función para borrar investigador
# -----------------------------------------
def borrar_investigador():

    listar_investigadores()

    print("\n--- BORRAR INVESTIGADOR ---")

    id_borrar = int(input("Ingresa el ID del investigador a borrar: "))

    for investigador in diccionario_investigadores:

        if investigador["ID"] == id_borrar:

            diccionario_investigadores.remove(investigador)

            print("\n🗑️ Investigador eliminado correctamente.")

            return

    print("\n❌ ID no encontrado.")


# -----------------------------------------
# Función para actualizar investigador
# -----------------------------------------
def actualizar_investigador():

    listar_investigadores()

    print("\n--- ACTUALIZAR INVESTIGADOR ---")

    id_actualizar = int(input("Ingresa el ID del investigador: "))

    for investigador in diccionario_investigadores:

        if investigador["ID"] == id_actualizar:

            investigador["Nombre"] = input("Nuevo nombre: ")
            investigador["Linea"] = input("Nueva línea de investigación: ")
            investigador["Publicaciones"] = input("Nuevas publicaciones: ")
            investigador["Universidad"] = input("Nueva universidad: ")
            investigador["Becas"] = input("Nuevas becas: ")
            investigador["SNI"] = input("Nuevo nivel SNI: ")
            investigador["Laboratorio"] = input("Nuevo laboratorio: ")

            print("\n📝 Investigador actualizado correctamente.")

            return

    print("\n❌ ID no encontrado.")


# -----------------------------------------
# MENÚ PRINCIPAL
# -----------------------------------------
def menu():

    while True:

        print("\n===================================")
        print("      SISTEMA DE INVESTIGADORES")
        print("===================================")

        print("1. Agregar investigador")
        print("2. Listar investigadores")
        print("3. Buscar investigador")
        print("4. Borrar investigador")
        print("5. Actualizar investigador")
        print("6. Salir")

        opcion = input("\nSelecciona una opción: ")

        if opcion == "1":
            agregar_investigador()

        elif opcion == "2":
            listar_investigadores()

        elif opcion == "3":
            buscar_investigador()

        elif opcion == "4":
            borrar_investigador()

        elif opcion == "5":
            actualizar_investigador()

        elif opcion == "6":
            print("\n👋 Saliendo del sistema...")
            break

        else:
            print("\n❌ Opción no válida.")


# -----------------------------------------
# INICIAR PROGRAMA
# -----------------------------------------
if __name__ == "__main__":
    menu()
```
