### Nivel: principiante puedes generar un programa en python que simule inteligencia artificial utilizando if. if else, elif, y while, referido a xxxx y xxxx tipo menu, que sea interactivo. No utilizar programación orientada a objetos, no funciones, no ciclos for, no try, el ejemplo debe ser simple y atractivo para alumnos de preparatoria

```python
# =========================================
# SISTEMA IA DE LOCUTOR Y MICROFONO
# Nivel: Principiante
# =========================================

opcion = 0

while opcion != 5:

    print("\n====================================")
    print("     SISTEMA IA DE RADIO")
    print("====================================")

    print("1. Analizar voz del locutor")
    print("2. Revisar volumen del microfono")
    print("3. Revisar energia del locutor")
    print("4. Analizar tipo de microfono")
    print("5. Salir")

    opcion = int(input("\nSelecciona una opcion: "))

    # =====================================
    # OPCION 1
    # =====================================

    if opcion == 1:

        print("\n====== ANALISIS DE VOZ ======")

        volumen = int(input("Nivel de voz del locutor: "))
        claridad = int(input("Nivel de claridad de voz: "))

        if volumen > 80 and claridad > 80:

            print("IA: VOZ PERFECTA PARA TRANSMISION")

        elif volumen > 60 and claridad > 60:

            print("IA: VOZ ESTABLE")

        elif volumen > 40 and claridad > 40:

            print("IA: VOZ BAJA")

        else:

            print("IA: VOZ DEFICIENTE")

    # =====================================
    # OPCION 2
    # =====================================

    elif opcion == 2:

        print("\n====== REVISION DE MICROFONO ======")

        sonido = int(input("Nivel de sonido del microfono: "))

        if sonido >= 90:

            print("IA: MICROFONO EN EXCELENTE ESTADO")

        elif sonido >= 70:

            print("IA: MICROFONO FUNCIONANDO CORRECTAMENTE")

        elif sonido >= 40:

            print("IA: MICROFONO CON FALLAS MENORES")

        else:

            print("IA: MICROFONO DAÑADO")

    # =====================================
    # OPCION 3
    # =====================================

    elif opcion == 3:

        print("\n====== ENERGIA DEL LOCUTOR ======")

        energia = int(input("Nivel de energia del locutor: "))

        if energia >= 80:

            print("IA: LOCUTOR MUY ACTIVO")

        elif energia >= 50:

            print("IA: LOCUTOR ESTABLE")

        elif energia >= 20:

            print("IA: LOCUTOR CANSADO")

        else:

            print("IA: DESCANSO NECESARIO")

    # =====================================
    # OPCION 4
    # =====================================

    elif opcion == 4:

        print("\n====== ANALISIS DE MICROFONO ======")

        microfono = input("Tipo de microfono (condensador/dinamico/inalambrico): ")

        if microfono == "condensador":

            print("IA: MICROFONO IDEAL PARA ESTUDIO")

        elif microfono == "dinamico":

            print("IA: MICROFONO RESISTENTE Y POTENTE")

        elif microfono == "inalambrico":

            print("IA: MICROFONO PRACTICO PARA EVENTOS")

        else:

            print("IA: MICROFONO DESCONOCIDO")

    # =====================================
    # OPCION 5
    # =====================================

    elif opcion == 5:

        print("\nSistema finalizado")

    # =====================================
    # OPCION INVALIDA
    # =====================================

    else:

        print("\nOpcion no valida")
```
