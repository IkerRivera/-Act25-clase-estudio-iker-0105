## militar y armas

```python
# =========================================
# SISTEMA IA MILITAR Y ARMAS
# Nivel: Principiante
# =========================================

opcion = 0

while opcion != 5:

    print("\n====================================")
    print("     SISTEMA IA MILITAR")
    print("====================================")

    print("1. Analizar combate")
    print("2. Revisar municion")
    print("3. Revisar energia")
    print("4. Analizar arma")
    print("5. Salir")

    opcion = int(input("\nSelecciona una opcion: "))

    # =====================================
    # OPCION 1
    # =====================================

    if opcion == 1:

        print("\n====== ANALISIS DE COMBATE ======")

        enemigos = int(input("Cantidad de enemigos: "))
        municion = int(input("Municion disponible: "))

        if enemigos > 50 and municion > 80:

            print("IA: ATAQUE TOTAL AUTORIZADO")

        elif enemigos > 20 and municion > 40:

            print("IA: COMBATE DIRECTO")

        elif enemigos > 10 and municion < 30:

            print("IA: RETIRADA ESTRATEGICA")

        else:

            print("IA: ZONA SEGURA")

    # =====================================
    # OPCION 2
    # =====================================

    elif opcion == 2:

        print("\n====== REVISION DE MUNICION ======")

        municion = int(input("Cantidad de municion: "))

        if municion >= 100:

            print("IA: MUNICION EXCELENTE")

        elif municion >= 50:

            print("IA: MUNICION ESTABLE")

        elif municion >= 20:

            print("IA: MUNICION BAJA")

        else:

            print("IA: MUNICION CRITICA")

    # =====================================
    # OPCION 3
    # =====================================

    elif opcion == 3:

        print("\n====== REVISION DE ENERGIA ======")

        energia = int(input("Nivel de energia (%): "))

        if energia >= 80:

            print("IA: SOLDADO EN PERFECTO ESTADO")

        elif energia >= 50:

            print("IA: SOLDADO ESTABLE")

        elif energia >= 20:

            print("IA: SOLDADO CANSADO")

        else:

            print("IA: DESCANSO OBLIGATORIO")

    # =====================================
    # OPCION 4
    # =====================================

    elif opcion == 4:

        print("\n====== ANALISIS DE ARMAS ======")

        arma = input("Tipo de arma (rifle/sniper/tanque): ")

        if arma == "rifle":

            print("IA: ARMA BALANCEADA PARA COMBATE")

        elif arma == "sniper":

            print("IA: IDEAL PARA ATAQUES A DISTANCIA")

        elif arma == "tanque":

            print("IA: POTENCIA DE FUEGO PESADA")

        else:

            print("IA: ARMA DESCONOCIDA")

    # =====================================
    # OPCION 5
    # =====================================

    elif opcion == 5:

        print("\nSistema finalizado")

    # =====================================
    # OPCION INVALIDA
    # =====================================

    else:

        print("\nOpcion no valida")
```
