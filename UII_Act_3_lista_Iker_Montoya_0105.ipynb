38. pescador

```python
# Base de datos volátil (en memoria)
inventario_pesca = []

def agregarPez():
    print("\n--- Registrar Nueva Captura ---")
    id_pez = len(inventario_pesca) + 1
    especie = input("Especie del pez: ")
    peso = float(input("Peso (kg): "))
    pescador = input("Nombre del pescador: ")

    nuevo_pez = {
        "id": id_pez,
        "especie": especie,
        "peso": peso,
        "pescador": pescador
    }
    inventario_pesca.append(nuevo_pez)
    print(f"✅ {especie} registrado con éxito.")

def listarPeces():
    print("\n--- Lista de Capturas del Día ---")
    if not inventario_pesca:
        print("La red está vacía. No hay capturas registradas.")
    else:
        for p in inventario_pesca:
            print(f"ID: {p['id']} | Especie: {p['especie']} | Peso: {p['peso']}kg | Pescador: {p['pescador']}")

def buscarPez():
    print("\n--- Buscar Captura ---")
    busqueda = input("Ingresa el nombre de la especie a buscar: ").lower()
    encontrados = [p for p in inventario_pesca if busqueda in p['especie'].lower()]

    if encontrados:
        for p in encontrados:
            print(f"🔎 Encontrado -> ID: {p['id']} | Especie: {p['especie']} | Pescador: {p['pescador']}")
    else:
        print("❌ No se encontró ningún pez con ese nombre.")

def modificarPez():
    listarPeces()
    id_mod = int(input("\nIngresa el ID del registro que deseas modificar: "))
    for p in inventario_pesca:
        if p['id'] == id_mod:
            p['especie'] = input(f"Nueva especie ({p['especie']}): ") or p['especie']
            p['peso'] = float(input(f"Nuevo peso ({p['peso']}kg): ") or p['peso'])
            print("📝 Registro actualizado.")
            return
    print("❌ ID no encontrado.")

def borrarPez():
    listarPeces()
    id_del = int(input("\nIngresa el ID del pez que se escapó (borrar): "))
    global inventario_pesca
    original_count = len(inventario_pesca)
    inventario_pesca = [p for p in inventario_pesca if p['id'] != id_del]

    if len(inventario_pesca) < original_count:
        print("🗑️ Registro eliminado de la bitácora.")
    else:
        print("❌ No se encontró el ID.")

def menu():
    while True:
        print("\n--- BITÁCORA DE PESCA ---")
        print("1. Agregar Pez")
        print("2. Listar Peces")
        print("3. Buscar Pez")
        print("4. Modificar Pez")
        print("5. Borrar Pez")
        print("6. Salir")

        opcion = input("Selecciona una opción: ")

        if opcion == "1": agregarPez()
        elif opcion == "2": listarPeces()
        elif opcion == "3": buscarPez()
        elif opcion == "4": modificarPez()
        elif opcion == "5": borrarPez()
        elif opcion == "6": break
        else: print("Opción no válida.")

# Iniciar el programa
if __name__ == "__main__":
    menu()
```
