#### puedes generar un programa en python que simule inteligencia artificial utilizando if. if else, elif, y operadores,referido a militar y armas, que sea interactivo. No utilizar programacion orientada a objetos, no funciones, no ciclos de repeticion, solamente los tipos de if condicional.


```python
# =========================================
# SISTEMA MILITAR CON INTELIGENCIA ARTIFICIAL
# Uso de if, elif, else y operadores
# =========================================

print("========================================")
print("   SISTEMA DE ANALISIS MILITAR IA")
print("========================================")

nombre = input("Nombre del soldado: ")
rango = input("Rango militar: ")

enemigos = int(input("Cantidad de enemigos detectados: "))
municion = int(input("Cantidad de municion disponible: "))
energia = int(input("Nivel de energia del soldado (%): "))
clima = input("Clima actual (soleado/lluvia/tormenta): ")

print("\n========================================")
print("      ANALISIS DE INTELIGENCIA")
print("========================================")

# -----------------------------------------
# DECISIONES DE LA IA
# -----------------------------------------

if enemigos >= 50 and municion >= 80 and energia >= 70:

    print("IA Militar: ATAQUE TOTAL")
    print("El soldado tiene suficientes recursos.")
    print("Mision autorizada.")

elif enemigos >= 50 and municion < 50:

    print("IA Militar: RETIRADA ESTRATEGICA")
    print("Municion insuficiente para combate.")

elif enemigos >= 20 and energia < 40:

    print("IA Militar: DESCANSO OBLIGATORIO")
    print("El soldado presenta fatiga extrema.")

elif clima == "tormenta":

    print("IA Militar: OPERACION CANCELADA")
    print("Condiciones climaticas peligrosas.")

elif enemigos < 10 and municion > 30:

    print("IA Militar: MISION DE RECONOCIMIENTO")
    print("Zona considerada segura.")

else:

    print("IA Militar: ESPERANDO NUEVAS ORDENES")
    print("Situacion bajo observacion.")

# -----------------------------------------
# RESULTADO FINAL
# -----------------------------------------

print("\n========================================")
print("          REPORTE FINAL")
print("========================================")

print("Soldado:", nombre)
print("Rango:", rango)
print("Enemigos:", enemigos)
print("Municion:", municion)
print("Energia:", energia)
print("Clima:", clima)

print("\nSistema finalizado.")
```
