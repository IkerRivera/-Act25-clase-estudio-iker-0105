```python
# =========================================
# SISTEMA IA MILITAR CON LISTAS
# Nivel: Principiante
# =========================================

# Lista vacia
soldados = []

opcion = 0

while opcion != 5:

    print("\n====================================")
    print("      SISTEMA IA MILITAR")
    print("====================================")

    print("1. Registrar soldado")
    print("2. Ver soldados")
    print("3. Analizar combate")
    print("4. Estado militar")
    print("5. Salir")

    opcion = int(input("\nSelecciona una opcion: "))

    # =====================================
    # OPCION 1
    # =====================================

    if opcion == 1:

        print("\n====== REGISTRO MILITAR ======")

        nombre = input("Nombre del soldado: ")
        rango = input("Rango militar: ")
        arma = input("Arma principal: ")

        datos = [nombre, rango, arma]

        soldados.append(datos)

        print("\nIA: Soldado registrado correctamente")

    # =====================================
    # OPCION 2
    # =====================================

    elif opcion == 2:

        print("\n============== LISTA DE SOLDADOS ==============")

        if soldados == []:

            print("\nIA: No hay soldados registrados")

        else:

            print("\n-----------------------------------------------")
            print("NOMBRE       | RANGO        | ARMA")
            print("-----------------------------------------------")

            i = 0

            while i < len(soldados):

                print(
                    soldados[i][0], " " * (12 - len(soldados[i][0])),
                    "|",
                    soldados[i][1], " " * (12 - len(soldados[i][1])),
                    "|",
                    soldados[i][2]
                )

                i = i + 1

            print("-----------------------------------------------")

    # =====================================
    # OPCION 3
    # =====================================

    elif opcion == 3:

        print("\n====== ANALISIS DE COMBATE ======")

        enemigos = int(input("Cantidad de enemigos: "))
        municion = int(input("Cantidad de municion: "))

        if enemigos > 50 and municion > 80:

            print("IA: ATAQUE TOTAL AUTORIZADO")

        elif enemigos > 20 and municion > 40:

            print("IA: COMBATE DIRECTO")

        elif enemigos > 10 and municion < 30:

            print("IA: RETIRADA ESTRATEGICA")

        else:

            print("IA: ZONA SEGURA")

    # =====================================
    # OPCION 4
    # =====================================

    elif opcion == 4:

        print("\n====== ESTADO MILITAR ======")

        energia = int(input("Nivel de energia del soldado: "))

        if energia >= 80:

            print("IA: SOLDADO EN PERFECTO ESTADO")

        elif energia >= 50:

            print("IA: SOLDADO ESTABLE")

        elif energia >= 20:

            print("IA: SOLDADO CANSADO")

        else:

            print("IA: DESCANSO OBLIGATORIO")

    # =====================================
    # OPCION 5
    # =====================================

    elif opcion == 5:

        print("\nSistema finalizado")

    # =====================================
    # OPCION INVALIDA
    # =====================================

    else:

        print("\nOpcion no valida")
```
