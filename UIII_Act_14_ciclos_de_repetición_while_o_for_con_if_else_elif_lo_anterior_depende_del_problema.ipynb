#Redacción del problema Numero = 42 Programa FizzBuzz
* Crear un programa en Python que imprima números del 1 al N.
* Si el número es múltiplo de 3 debe mostrar "Fizz".
* Si el número es múltiplo de 5 debe mostrar "Buzz".
* Si el número es múltiplo de ambos debe mostrar "FizzBuzz".


```python
# =========================================
# PROGRAMA FIZZBUZZ
# Nivel: Principiante
# =========================================

print("================================")
print("        PROGRAMA FIZZBUZZ")
print("================================")

# Entrada de datos
numero = int(input("Ingresa el numero final: "))

# Contador
i = 1

# Ciclo while
while i <= numero:

    # Multiplo de 3 y 5
    if i % 3 == 0 and i % 5 == 0:

        print("FizzBuzz")

    # Multiplo de 3
    elif i % 3 == 0:

        print("Fizz")

    # Multiplo de 5
    elif i % 5 == 0:

        print("Buzz")

    # Numero normal
    else:

        print(i)

    i = i + 1
```
