```python
# =========================================
# SISTEMA IA DE MUSICOS
# PROGRAMACION ORIENTADA A OBJETOS
# Nivel: Principiante
# =========================================


# =========================================
# CLASE MUSICO
# =========================================
class Musico:

    # Constructor
    def __init__(self, nombre, instrumento, anios,
                 banda, partituras, conciertos,
                 conservatorio):

        self.nombre = nombre
        self.instrumento = instrumento
        self.anios = anios
        self.banda = banda
        self.partituras = partituras
        self.conciertos = conciertos
        self.conservatorio = conservatorio


# =========================================
# LISTA DE MUSICOS
# =========================================
musicos = []


# =========================================
# FUNCION AGREGAR MUSICO
# =========================================
def agregar_musico():

    print("\n========== REGISTRO DE MUSICO ==========")

    nombre = input("Nombre del musico: ")
    instrumento = input("Instrumento principal: ")
    anios = input("Años de estudio: ")
    banda = input("Orquesta o banda: ")
    partituras = input("Partituras compuestas: ")
    conciertos = input("Conciertos al año: ")
    conservatorio = input("Conservatorio: ")

    nuevo_musico = Musico(
        nombre,
        instrumento,
        anios,
        banda,
        partituras,
        conciertos,
        conservatorio
    )

    musicos.append(nuevo_musico)

    print("\nIA: Musico registrado correctamente")


# =========================================
# FUNCION LISTAR MUSICOS
# =========================================
def listar_musico():

    print("\n==================== LISTA DE MUSICOS ====================")

    if musicos == []:

        print("\nNo hay musicos registrados")

    else:

        print("----------------------------------------------------------------------------------------------------------------")
        print("ID | NOMBRE        | INSTRUMENTO    | AÑOS | BANDA           | PARTITURAS | CONCIERTOS | CONSERVATORIO")
        print("----------------------------------------------------------------------------------------------------------------")

        contador = 1

        for musico in musicos:

            print(
                f"{contador:<2} | "
                f"{musico.nombre:<13} | "
                f"{musico.instrumento:<14} | "
                f"{musico.anios:<5} | "
                f"{musico.banda:<15} | "
                f"{musico.partituras:<11} | "
                f"{musico.conciertos:<11} | "
                f"{musico.conservatorio}"
            )

            contador = contador + 1

        print("----------------------------------------------------------------------------------------------------------------")


# =========================================
# FUNCION ACTUALIZAR MUSICO
# =========================================
def actualizar_musico():

    listar_musico()

    if musicos != []:

        id_musico = int(input("\nIngresa el ID del musico a actualizar: "))

        if id_musico >= 1 and id_musico <= len(musicos):

            musicos[id_musico - 1].nombre = input("Nuevo nombre: ")
            musicos[id_musico - 1].instrumento = input("Nuevo instrumento: ")
            musicos[id_musico - 1].anios = input("Nuevos años de estudio: ")
            musicos[id_musico - 1].banda = input("Nueva banda: ")
            musicos[id_musico - 1].partituras = input("Nuevas partituras: ")
            musicos[id_musico - 1].conciertos = input("Nuevos conciertos: ")
            musicos[id_musico - 1].conservatorio = input("Nuevo conservatorio: ")

            print("\nIA: Musico actualizado correctamente")

        else:

            print("\nID no encontrado")


# =========================================
# FUNCION BORRAR MUSICO
# =========================================
def borrar_musico():

    listar_musico()

    if musicos != []:

        id_musico = int(input("\nIngresa el ID del musico a borrar: "))

        if id_musico >= 1 and id_musico <= len(musicos):

            musicos.pop(id_musico - 1)

            print("\nIA: Musico eliminado correctamente")

        else:

            print("\nID no encontrado")


# =========================================
# MENU PRINCIPAL
# =========================================
while True:

    print("\n========================================")
    print("        SISTEMA IA DE MUSICOS")
    print("========================================")

    print("1. Agregar musico")
    print("2. Listar musicos")
    print("3. Actualizar musico")
    print("4. Borrar musico")
    print("5. Salir")

    opcion = input("\nSelecciona una opcion: ")

    # =====================================
    # OPCION 1
    # =====================================

    if opcion == "1":

        agregar_musico()

    # =====================================
    # OPCION 2
    # =====================================

    elif opcion == "2":

        listar_musico()

    # =====================================
    # OPCION 3
    # =====================================

    elif opcion == "3":

        actualizar_musico()

    # =====================================
    # OPCION 4
    # =====================================

    elif opcion == "4":

        borrar_musico()

    # =====================================
    # OPCION 5
    # =====================================

    elif opcion == "5":

        print("\nSistema finalizado")
        break

    # =====================================
    # OPCION INVALIDA
    # =====================================

    else:

        print("\nOpcion no valida")
```
