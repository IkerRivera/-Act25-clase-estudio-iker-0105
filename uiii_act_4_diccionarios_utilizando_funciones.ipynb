## rol : Investigador: Línea de investigación, publicaciones científicas, universidad, becas obtenidas, nivel de SNI y laboratorio.

```python
# Base de datos volátil (en memoria)
diccionario_investigadores = {}


# =========================================
# AGREGAR INVESTIGADOR
# =========================================
def agregar_investigador():

    print("\n--- Registrar Investigador ---")

    id_investigador = len(diccionario_investigadores) + 1

    nombre = input("Nombre del investigador: ")
    linea = input("Línea de investigación: ")
    publicaciones = input("Publicaciones científicas: ")
    universidad = input("Universidad: ")
    becas = input("Becas obtenidas: ")
    sni = input("Nivel de SNI: ")
    laboratorio = input("Laboratorio: ")

    nuevo_investigador = {
        "id": id_investigador,
        "nombre": nombre,
        "linea": linea,
        "publicaciones": publicaciones,
        "universidad": universidad,
        "becas": becas,
        "sni": sni,
        "laboratorio": laboratorio
    }

    diccionario_investigadores.append(nuevo_investigador)

    print("\n✅ Investigador registrado correctamente.")


# =========================================
# LISTAR INVESTIGADORES
# =========================================
def listar_investigadores():

    print("\n================ TABLA DE INVESTIGADORES ================\n")

    if not diccionario_investigadores:

        print("No hay investigadores registrados.")

    else:

        print("{:<4} {:<18} {:<20} {:<18} {:<18} {:<15} {:<10} {:<15}".format(
            "ID",
            "NOMBRE",
            "LINEA",
            "PUBLICACIONES",
            "UNIVERSIDAD",
            "BECAS",
            "SNI",
            "LABORATORIO"
        ))

        print("-" * 130)

        for i in diccionario_investigadores:

            print("{:<4} {:<18} {:<20} {:<18} {:<18} {:<15} {:<10} {:<15}".format(
                i["id"],
                i["nombre"],
                i["linea"],
                i["publicaciones"],
                i["universidad"],
                i["becas"],
                i["sni"],
                i["laboratorio"]
            ))


# =========================================
# BUSCAR INVESTIGADOR
# =========================================
def buscar_investigador():

    listar_investigadores()

    print("\n--- Buscar Investigador ---")

    busqueda = input("Ingresa el nombre del investigador: ").lower()

    encontrados = [
        i for i in diccionario_investigadores
        if busqueda in i["nombre"].lower()
    ]

    if encontrados:

        print("\n=========== RESULTADOS ===========\n")

        print("{:<4} {:<18} {:<20} {:<18}".format(
            "ID",
            "NOMBRE",
            "LINEA",
            "UNIVERSIDAD"
        ))

        print("-" * 70)

        for i in encontrados:

            print("{:<4} {:<18} {:<20} {:<18}".format(
                i["id"],
                i["nombre"],
                i["linea"],
                i["universidad"]
            ))

    else:
        print("\n❌ Investigador no encontrado.")


# =========================================
# MODIFICAR INVESTIGADOR
# =========================================
def actualizar_investigador():

    listar_investigadores()

    id_mod = int(input("\nIngresa el ID que deseas modificar: "))

    for i in diccionario_investigadores:

        if i["id"] == id_mod:

            i["nombre"] = input(f"Nuevo nombre ({i['nombre']}): ") or i["nombre"]

            i["linea"] = input(f"Nueva línea ({i['linea']}): ") or i["linea"]

            i["publicaciones"] = input(f"Nuevas publicaciones ({i['publicaciones']}): ") or i["publicaciones"]

            i["universidad"] = input(f"Nueva universidad ({i['universidad']}): ") or i["universidad"]

            i["becas"] = input(f"Nuevas becas ({i['becas']}): ") or i["becas"]

            i["sni"] = input(f"Nuevo nivel SNI ({i['sni']}): ") or i["sni"]

            i["laboratorio"] = input(f"Nuevo laboratorio ({i['laboratorio']}): ") or i["laboratorio"]

            print("\n📝 Registro actualizado correctamente.")

            return

    print("\n❌ ID no encontrado.")


# =========================================
# BORRAR INVESTIGADOR
# =========================================
def borrar_investigador():

    listar_investigadores()

    id_del = int(input("\nIngresa el ID a borrar: "))

    global diccionario_investigadores

    cantidad_original = len(diccionario_investigadores)

    diccionario_investigadores = [
        i for i in diccionario_investigadores
        if i["id"] != id_del
    ]

    if len(diccionario_investigadores) < cantidad_original:

        print("\n🗑️ Registro eliminado correctamente.")

    else:
        print("\n❌ ID no encontrado.")


# =========================================
# MENÚ PRINCIPAL
# =========================================
def menu():

    while True:

        print("\n========== SISTEMA DE INVESTIGADORES ==========")

        print("1. Agregar investigador")
        print("2. Listar investigadores")
        print("3. Buscar investigador")
        print("4. Actualizar investigador")
        print("5. Borrar investigador")
        print("6. Salir")

        opcion = input("\nSelecciona una opción: ")

        if opcion == "1":
            agregar_investigador()

        elif opcion == "2":
            listar_investigadores()

        elif opcion == "3":
            buscar_investigador()

        elif opcion == "4":
            actualizar_investigador()

        elif opcion == "5":
            borrar_investigador()

        elif opcion == "6":
            print("\n👋 Programa finalizado.")
            break

        else:
            print("\n❌ Opción no válida.")


# Iniciar programa
if __name__ == "__main__":
    menu()
```
