```python
# =========================================
# SISTEMA IA TARJETA PREPAGADA
# PROGRAMACION ORIENTADA A OBJETOS
# Nivel: Principiante
# =========================================


# =========================================
# CLASE TARJETA PREPAGADA
# =========================================
class Tarjeta_Prepagada:

    # Constructor
    def __init__(self, numero_tarjeta, saldo,
                 usuario, fecha_vencimiento,
                 estado):

        self.numero_tarjeta = numero_tarjeta
        self.saldo = saldo
        self.usuario = usuario
        self.fecha_vencimiento = fecha_vencimiento
        self.estado = estado

    # =====================================
    # FUNCION RECARGAR SALDO
    # =====================================
    def recargar_saldo(self):

        cantidad = float(input("Cantidad a recargar: "))

        self.saldo = self.saldo + cantidad

        print("\nIA: Recarga realizada correctamente")

    # =====================================
    # FUNCION COMPRAR
    # =====================================
    def comprar(self):

        if self.estado == "Bloqueada":

            print("\nIA: Tarjeta suspendida")

        else:

            compra = float(input("Cantidad de compra: "))

            if compra <= self.saldo:

                self.saldo = self.saldo - compra

                print("\nIA: Compra realizada")

            else:

                print("\nIA: Saldo insuficiente")

    # =====================================
    # FUNCION CONSULTAR SALDO
    # =====================================
    def consultar_saldo(self):

        print("\n========== SALDO ==========")
        print("Saldo disponible:", self.saldo)

        if self.saldo >= 1000:

            print("IA: Saldo excelente")

        elif self.saldo >= 500:

            print("IA: Saldo estable")

        elif self.saldo >= 100:

            print("IA: Recarga recomendada")

        else:

            print("IA: Saldo muy bajo")

    # =====================================
    # FUNCION BLOQUEAR TARJETA
    # =====================================
    def bloquear_tarjeta(self):

        self.estado = "Bloqueada"

        print("\nIA: Tarjeta bloqueada")

    # =====================================
    # FUNCION MOSTRAR DATOS
    # =====================================
    def mostrar_datos(self):

        print("--------------------------------------------------------------------------------")
        print("NUMERO       | USUARIO       | SALDO   | VENCIMIENTO   | ESTADO")
        print("--------------------------------------------------------------------------------")

        print(
            f"{self.numero_tarjeta:<12} | "
            f"{self.usuario:<13} | "
            f"{self.saldo:<8} | "
            f"{self.fecha_vencimiento:<13} | "
            f"{self.estado}"
        )

        print("--------------------------------------------------------------------------------")


# =========================================
# CREAR OBJETO
# =========================================

print("========================================")
print("     REGISTRO TARJETA PREPAGADA")
print("========================================")

numero_tarjeta = input("Numero de tarjeta: ")
saldo = float(input("Saldo inicial: "))
usuario = input("Nombre del usuario: ")
fecha_vencimiento = input("Fecha de vencimiento: ")
estado = input("Estado de la tarjeta: ")

tarjeta = Tarjeta_Prepagada(
    numero_tarjeta,
    saldo,
    usuario,
    fecha_vencimiento,
    estado
)


# =========================================
# MENU PRINCIPAL
# =========================================
while True:

    print("\n========================================")
    print("     SISTEMA IA TARJETA PREPAGADA")
    print("========================================")

    print("1. Recargar saldo")
    print("2. Comprar")
    print("3. Consultar saldo")
    print("4. Bloquear tarjeta")
    print("5. Mostrar datos")
    print("6. Salir")

    opcion = input("\nSelecciona una opcion: ")

    # =====================================
    # OPCION 1
    # =====================================

    if opcion == "1":

        tarjeta.recargar_saldo()

    # =====================================
    # OPCION 2
    # =====================================

    elif opcion == "2":

        tarjeta.comprar()

    # =====================================
    # OPCION 3
    # =====================================

    elif opcion == "3":

        tarjeta.consultar_saldo()

    # =====================================
    # OPCION 4
    # =====================================

    elif opcion == "4":

        tarjeta.bloquear_tarjeta()

    # =====================================
    # OPCION 5
    # =====================================

    elif opcion == "5":

        tarjeta.mostrar_datos()

    # =====================================
    # OPCION 6
    # =====================================

    elif opcion == "6":

        print("\nSistema finalizado")
        break

    # =====================================
    # OPCION INVALIDA
    # =====================================

    else:

        print("\nOpcion no valida")
```
