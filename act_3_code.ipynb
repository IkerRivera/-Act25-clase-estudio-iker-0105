```python
# Sistema básico de inicio de sesión usando funciones (def)

# Variables globales
usuario_guardado = ""
contrasena_guardada = ""


# Función para registrarse
def registrarse():
    global usuario_guardado, contrasena_guardada

    usuario_guardado = input("Crea un usuario: ")
    contrasena_guardada = input("Crea una contraseña: ")

    print("Registro exitoso")


# Función para iniciar sesión
def iniciar_sesion():
    usuario = input("Ingresa tu usuario: ")
    contrasena = input("Ingresa tu contraseña: ")

    if usuario == usuario_guardado and contrasena == contrasena_guardada:
        print("Inicio de sesión exitoso")
    else:
        print("Usuario o contraseña incorrectos")


# Función para mostrar el menú
def menu():
    print("\n--- MENÚ PRINCIPAL ---")
    print("1. Registrarse")
    print("2. Iniciar sesión")
    print("3. Salir")


# Programa principal
opcion = 0

while opcion != 3:

    menu()

    opcion = int(input("Selecciona una opción: "))

    if opcion == 1:
        registrarse()

    elif opcion == 2:
        iniciar_sesion()

    elif opcion == 3:
        print("Saliendo del sistema...")

    else:
        print("Opción no válida")
```
