```python
# =========================================
# SISTEMA IA MILITAR
# DICCIONARIOS
# Nivel: Principiante
# =========================================

# Diccionario vacio
militar = {}

# Diccionario vacio principal
los_militares = {}

contador = 1

while True:

    print("\n========================================")
    print("         SISTEMA IA MILITAR")
    print("========================================")

    print("1. Registrar militar")
    print("2. Ver militares")
    print("3. Analizar combate")
    print("4. Revisar energia")
    print("5. Salir")

    opcion = int(input("\nSelecciona una opcion: "))

    # =====================================
    # OPCION 1
    # =====================================

    if opcion == 1:

        print("\n========== REGISTRO MILITAR ==========")

        militar = {}

        nombre = input("Nombre del militar: ")
        rango = input("Rango militar: ")
        arma = input("Arma principal: ")
        energia = input("Nivel de energia: ")

        militar["Nombre"] = nombre
        militar["Rango"] = rango
        militar["Arma"] = arma
        militar["Energia"] = energia

        los_militares[contador] = militar

        contador = contador + 1

        print("\nIA: Militar registrado correctamente")

    # =====================================
    # OPCION 2
    # =====================================

    elif opcion == 2:

        print("\n============== LISTA DE MILITARES ==============")

        if los_militares == {}:

            print("\nNo hay militares registrados")

        else:

            print("------------------------------------------------------------")
            print("ID | NOMBRE        | RANGO        | ARMA       | ENERGIA")
            print("------------------------------------------------------------")

            for clave in los_militares:

                print(
                    f"{clave:<2} | "
                    f"{los_militares[clave]['Nombre']:<13} | "
                    f"{los_militares[clave]['Rango']:<13} | "
                    f"{los_militares[clave]['Arma']:<11} | "
                    f"{los_militares[clave]['Energia']}"
                )

            print("------------------------------------------------------------")

    # =====================================
    # OPCION 3
    # =====================================

    elif opcion == 3:

        print("\n========== ANALISIS DE COMBATE ==========")

        enemigos = int(input("Cantidad de enemigos: "))
        municion = int(input("Cantidad de municion: "))

        if enemigos > 50 and municion > 80:

            print("\nIA: ATAQUE TOTAL AUTORIZADO")

        elif enemigos > 20 and municion > 40:

            print("\nIA: COMBATE DIRECTO")

        elif enemigos > 10 and municion < 30:

            print("\nIA: RETIRADA ESTRATEGICA")

        else:

            print("\nIA: ZONA SEGURA")

    # =====================================
    # OPCION 4
    # =====================================

    elif opcion == 4:

        print("\n========== ESTADO MILITAR ==========")

        energia = int(input("Nivel de energia: "))

        if energia >= 80:

            print("\nIA: SOLDADO EN PERFECTO ESTADO")

        elif energia >= 50:

            print("\nIA: SOLDADO ESTABLE")

        elif energia >= 20:

            print("\nIA: SOLDADO CANSADO")

        else:

            print("\nIA: DESCANSO OBLIGATORIO")

    # =====================================
    # OPCION 5
    # =====================================

    elif opcion == 5:

        print("\nSistema finalizado")
        break

    # =====================================
    # OPCION INVALIDA
    # =====================================

    else:

        print("\nOpcion no valida")
```
