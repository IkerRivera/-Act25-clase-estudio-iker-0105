puedes generar un programa en python que simule inteligencia artificial utilizando if. if else, elif, y while,referido a xxxx, que sea interactivo. No utilizar programacion orientada a objetos, no funciones, no ciclos for,  no try,

```python
# =========================================
# SISTEMA IA DE TARJETA PREPAGADA
# Nivel: Principiante
# Uso de if, elif, else y while
# =========================================

continuar = "si"

while continuar == "si":

    print("\n======================================")
    print("     SISTEMA IA TARJETA PREPAGADA")
    print("======================================")

    nombre = input("Nombre del usuario: ")

    saldo = float(input("Saldo disponible: "))
    recarga = float(input("Cantidad de recarga: "))
    consumo = float(input("Consumo realizado: "))

    saldo_final = saldo + recarga - consumo

    print("\n======================================")
    print("       ANALISIS DE LA IA")
    print("======================================")

    # ==================================
    # DECISIONES DE LA IA
    # ==================================

    if saldo_final >= 500:

        print("IA: Saldo excelente")
        print("La tarjeta tiene suficiente credito.")

    elif saldo_final >= 200 and saldo_final < 500:

        print("IA: Saldo estable")
        print("Puedes seguir utilizando servicios.")

    elif saldo_final >= 50 and saldo_final < 200:

        print("IA: Saldo bajo")
        print("Se recomienda realizar una recarga.")

    elif saldo_final > 0 and saldo_final < 50:

        print("IA: Saldo muy bajo")
        print("Recarga urgente recomendada.")

    else:

        print("IA: Saldo agotado")
        print("La tarjeta ha sido suspendida temporalmente.")

    # ==================================
    # REPORTE FINAL
    # ==================================

    print("\n======================================")
    print("          REPORTE FINAL")
    print("======================================")

    print("Usuario:", nombre)
    print("Saldo inicial:", saldo)
    print("Recarga:", recarga)
    print("Consumo:", consumo)
    print("Saldo final:", saldo_final)

    continuar = input("\n¿Deseas ejecutar nuevamente el sistema? (si/no): ")

print("\nSistema finalizado.")
```
